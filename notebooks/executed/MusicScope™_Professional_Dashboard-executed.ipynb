```python
# Helper to render the Scenario Dials without crowding the notebook intro
def render_scenario_dials():
    """Build the slider playground for decision-ready breakout experimentation."""
    try:
        import ipywidgets as widgets
        from IPython.display import display, clear_output
        import plotly.express as px
        import pandas as pd
    except Exception as exc:
        print(f"ℹ️ ipywidgets/plotly unavailable: {exc}. Install ipywidgets for interactive controls.")
        return

    if 'momentum_daily' not in globals() or momentum_daily.empty:
        print("⚠️ Run the momentum scoring cell above before using these sliders (momentum_daily missing).")
        return

    pre_slider = widgets.IntSlider(value=55, min=40, max=70, step=1, description='Pre-breakout')
    brk_slider = widgets.IntSlider(value=60, min=50, max=85, step=1, description='Breakout')
    leg_slider = widgets.IntSlider(value=75, min=60, max=95, step=1, description='Legacy')
    slider_box = widgets.VBox([pre_slider, brk_slider, leg_slider])
    out = widgets.Output()

    def _artist_lookup() -> pd.DataFrame:
        """Return best-available mapping of video_id -> artist_name for attribution."""
        if 'videos_df' in globals() and not videos_df.empty:
            cols = [col for col in ['video_id', 'artist_name'] if col in videos_df.columns]
            if len(cols) == 2:
                return videos_df[cols].drop_duplicates()
        if 'momentum_daily' in globals() and {'video_id', 'artist_name'}.issubset(momentum_daily.columns):
            return momentum_daily[['video_id', 'artist_name']].drop_duplicates()
        return pd.DataFrame(columns=['video_id', 'artist_name'])

    artist_lookup = _artist_lookup()

    def _recompute(*_):
        with out:
            clear_output(wait=True)
            pre, brk, leg = pre_slider.value, brk_slider.value, leg_slider.value
            md = momentum_daily.copy()

            try:
                vbr = compute_kpi22_video_breakouts(md, pre=pre, brk=brk)
            except TypeError:
                vbr = compute_kpi22_video_breakouts(md, pre=pre)
            except Exception as exc:
                print(f"⚠️ Could not compute video breakouts: {exc}")
                return

            if vbr is None or vbr.empty:
                print(f"Breakout threshold {brk}: no qualifying video breakout periods yet.")
                return

            augmented = vbr.copy()
            if not artist_lookup.empty and 'video_id' in augmented.columns:
                augmented = augmented.merge(artist_lookup, on='video_id', how='left')
            if 'artist_name' not in augmented.columns:
                augmented['artist_name'] = '(unknown)'

            counts = (
                augmented.groupby('artist_name')
                         .size()
                         .rename('video_breakouts')
                         .reset_index()
                         .sort_values('video_breakouts', ascending=False)
            )

            print(f"Pre-breakout ≥{pre}, breakout ≥{brk}, legacy guardrail ≥{leg}")
            print(f"Breakout threshold {brk}: {len(vbr)} video breakout periods")

            if 'artist_name' in md.columns and 'momentum_score' in md.columns:
                legacy_summary = (
                    md.groupby('artist_name')['momentum_score']
                      .max()
                      .reset_index()
                      .rename(columns={'momentum_score': 'peak_score'})
                )
                cleared = legacy_summary[legacy_summary['peak_score'] >= leg]
                print(f"{len(cleared)} artists cleared the legacy guardrail (≥{leg}).")

            if counts.empty:
                print("Artist attribution missing—no chart to display.")
                return

            fig = px.bar(
                counts.head(15),
                x='video_breakouts',
                y='artist_name',
                orientation='h',
                title='Video Breakouts by Artist'
            )
            fig.update_layout(xaxis_title='Breakout periods', yaxis_title='')
            fig.show()

    for slider in (pre_slider, brk_slider, leg_slider):
        slider.observe(_recompute, names='value')

    display(slider_box, out)
    _recompute()
```

```python
# --- A6: KPI 22 — urgent double-panel -------------------------------------
import matplotlib.pyplot as plt
from matplotlib.dates import DateFormatter

# --- Compute END_DATE and episodes if not already defined ---
if 'END_DATE' not in globals():
    END_DATE = momentum_daily['date'].max()

if 'episodes' not in globals():
    # Detect breakout episodes: periods where momentum_score >= 60
    df = momentum_daily.copy()
    df = df.sort_values('date')
    df['is_breakout'] = (df['momentum_score'] >= 60).astype(int)
    # Find start/end of each episode
    df['br_shift'] = df['is_breakout'].shift(fill_value=0)
    starts = df[(df['is_breakout'] == 1) & (df['br_shift'] == 0)].index
    ends = df[(df['is_breakout'] == 0) & (df['br_shift'] == 1)].index

    # If a breakout is ongoing at the end, close it at the last date
    if len(starts) > len(ends):
        ends = ends.append(pd.Index([df.index[-1]]))

    episodes_list = []
    for s, e in zip(starts, ends):
        start_row = df.loc[s]
        end_row = df.loc[e]
        episode = {
            'start': start_row['date'],
            'end': end_row['date'],
            'title': start_row['title'],
            'days': (end_row['date'] - start_row['date']).days + 1,
            'pre_warning_hours': 0  # Placeholder, see below
        }
        episodes_list.append(episode)
    episodes = pd.DataFrame(episodes_list)
    # Compute pre-warning hours: count hours in pre-breakout state before each episode
    # (Assume daily granularity, so use days*24 as a proxy)
    for i, row in episodes.iterrows():
        pre_start = row['start'] - pd.Timedelta(days=2)
        pre_df = df[(df['date'] >= pre_start) & (df['date'] < row['start'])]
        pre_hours = (pre_df['momentum_score'].between(55, 60, inclusive='left')).sum() * 24
        episodes.at[i, 'pre_warning_hours'] = pre_hours

# Top-N recent episodes for the top panel
recent_eps = episodes[episodes["end"] >= (END_DATE - pd.Timedelta(days=30))].head(12)

# Aggregate daily breakout intensity in last 30 days for highlights
md = momentum_daily.assign(is_br=(momentum_daily["momentum_score"]>=60).astype(int))
last30 = pd.date_range(END_DATE - pd.Timedelta(days=30), END_DATE, freq="D")
hot = (md[md["date"].isin(last30)].groupby("date")["is_br"].sum()
       .sort_values(ascending=False).head(4))

fig, (ax1, ax2) = plt.subplots(2,1, figsize=(11,9))

# Panel 1 — Breakout duration (days)
ax1.barh(recent_eps["title"], recent_eps["days"], color="#D95F02")
for y, d in zip(recent_eps["title"], recent_eps["days"]):
    ax1.text(d, y, f" {int(d)}d", va="center", ha="left", fontsize=10, color="#222")
ax1.set_xlabel("Days in breakout (≥60)")
ax1.set_title("Breakout duration spiking on recent drops → shift budget now", fontweight="bold")

# Panel 2 — Pre-breakout warning hours
top_warn = episodes.sort_values("pre_warning_hours", ascending=False).head(12)
ax2.barh(top_warn["title"], top_warn["pre_warning_hours"], color="#1B9E77")
for y, h in zip(top_warn["title"], top_warn["pre_warning_hours"]):
    ax2.text(h, y, f" {int(h)}h", va="center", ha="left", fontsize=10, color="#222")
ax2.set_xlabel("Pre-breakout warning time (hours where 55≤score<60)")
ax2.set_title("Warning windows lengthening → monitor & seed before the jump", fontweight="bold")

# Highlight 3–4 hottest recent calendar days (annotation)
txt = "Hot days: " + ", ".join([d.strftime("%b %d %Y") for d in hot.index])
fig.suptitle("KPI 22 — Breakout duration & pre-breakout warning time", fontweight="bold")
fig.text(0.02, 0.01, txt, color="#D95F02", fontsize=10)
plt.tight_layout(); plt.show()
```

```python
# --- Interactive Threshold Experimentation (Analyst Playground) ---
try:
    import ipywidgets as widgets
    from IPython.display import display, clear_output
    import pandas as pd
    import plotly.express as px
except Exception as e:
    print(f"ℹ️ ipywidgets/plotly unavailable: {e}. Install ipywidgets for interactive controls.")
else:
    pre_slider = widgets.IntSlider(value=55, min=40, max=70, step=1, description='Pre-breakout')
    brk_slider = widgets.IntSlider(value=60, min=50, max=85, step=1, description='Breakout')
    leg_slider = widgets.IntSlider(value=75, min=60, max=95, step=1, description='Legacy')
    box = widgets.VBox([pre_slider, brk_slider, leg_slider])

    out = widgets.Output()

    def _recompute(*_):
        with out:
            clear_output(wait=True)
            pre, brk, leg = pre_slider.value, brk_slider.value, leg_slider.value
            md = momentum_daily.copy()
            try:
                vbr = compute_kpi22_video_breakouts(md, pre=pre, brk=brk)
            except Exception as e:
                print(f"⚠️ Could not compute video breakouts: {e}")
                return
            # summary counts
            n_breakouts = len(vbr)
            # per-artist counts
            if 'video_id' in md.columns and 'title' in md.columns:
                vids = md[['video_id']].drop_duplicates()
            # artist attribution (fallback by joining vids table if available in scope)
            try:
                per_artist = vbr.merge(vids, on='video_id', how='left')
            except Exception:
                per_artist = vbr.copy()
            if 'artist_name' in per_artist.columns:
                counts = per_artist.groupby('artist_name').size().rename('video_breakouts').reset_index()
            else:
                counts = per_artist.assign(artist_name='(unknown)').groupby('artist_name').size().rename('video_breakouts').reset_index()
            print(f"Breakout threshold {brk}: {n_breakouts} video breakout periods")
            # simple bar chart
            if not counts.empty:
                fig = px.bar(counts.sort_values('video_breakouts', ascending=False).head(15),
                             x='video_breakouts', y='artist_name', orientation='h', title='Video Breakouts by Artist')
                fig.show()

    pre_slider.observe(_recompute, names='value')
    brk_slider.observe(_recompute, names='value')
    leg_slider.observe(_recompute, names='value')
    display(box, out)
    _recompute()
```
