# 🎭 Sentiment Analysis & Fan Insights Dashboard

This notebook demonstrates **advanced sentiment analysis** with the specific charts requested:
- Divergent stacked bar charts for sentiment breakdown
- Sentiment cluster analysis showing model categories
- Top 3 positive/negative fan comments with percentages
- Standout video identification for experimentation
- Roster-wide sentiment analysis and tour grouping

**Focus: Compassionate analytics that treat artists as humans, not just data points** 💝

```python
# Import required modules
import pandas as pd
import numpy as np
from datetime import datetime, timedelta
import plotly.express as px
import plotly.graph_objects as go

# Import sentiment analysis functions
from youtubeviz.sentiment import (
    extract_top_positive_comments,
    extract_top_negative_comments_with_percentages,
    identify_standout_videos,
    analyze_roster_sentiment,
    group_artists_for_tours,
    detect_music_slang
)

# Import chart functions
from youtubeviz.charts import (
    create_divergent_sentiment_chart,
    create_sentiment_cluster_chart
)

# Import storytelling functions
from youtubeviz.storytelling import story_block

print('✅ Sentiment analysis toolkit loaded!')
```

```python
# Create sample sentiment data
artists = ['Flyana Boss', 'BiC Fizzle', 'COBRAH', 're6ce']

# Sample comments with realistic music industry language
positive_comments = [
    'This track is absolute fire! 🔥',
    'No cap, this is a banger',
    'Love the energy, this slaps hard',
    'Amazing vocals, gives me chills',
    'This goes so hard, on repeat',
    'Incredible beat, love the vibe'
]

negative_comments = [
    'Not feeling this one, mid tbh',
    'Could be better, not my style',
    'The beat is off, doesn\'t work',
    'Skip for me, not it'
]

# Generate sentiment data
sentiment_data = []
video_sentiment_data = []

for artist in artists:
    # Generate positive comments (more positive than negative)
    for i in range(np.random.randint(15, 25)):
        comment = np.random.choice(positive_comments)
        sentiment_data.append({
            'artist_name': artist,
            'comment': comment,
            'sentiment_score': np.random.uniform(0.6, 0.95),
            'sentiment_category': 'positive'
        })
    
    # Generate negative comments (fewer)
    for i in range(np.random.randint(3, 8)):
        comment = np.random.choice(negative_comments)
        sentiment_data.append({
            'artist_name': artist,
            'comment': comment,
            'sentiment_score': np.random.uniform(-0.8, -0.2),
            'sentiment_category': 'negative'
        })
    
    # Generate video sentiment data
    for i in range(np.random.randint(3, 6)):
        views = np.random.randint(10000, 150000)
        positive_pct = np.random.uniform(60, 90)
        
        video_sentiment_data.append({
            'artist_name': artist,
            'video_title': f'Video {i+1} - {artist}',
            'views': views,
            'positive_sentiment_pct': positive_pct,
            'negative_sentiment_pct': 100 - positive_pct
        })

sentiment_df = pd.DataFrame(sentiment_data)
video_df = pd.DataFrame(video_sentiment_data)

print(f'📊 Generated {len(sentiment_df)} comments for {len(artists)} artists')
print(f'🎬 Generated {len(video_df)} video sentiment records')
sentiment_df.head()
```

```python
# Create divergent stacked bar chart for sentiment breakdown
divergent_chart = create_divergent_sentiment_chart(
    df=sentiment_df,
    artist_col='artist_name',
    sentiment_col='sentiment_category',
    title='🎭 Fan Sentiment Breakdown by Artist'
)

story_block(
    fig=divergent_chart,
    title='💬 What Fans Are Really Saying',
    bullets=[
        'Green bars show positive fan sentiment (above baseline)',
        'Red bars show areas for improvement (below baseline)',
        'Chart reveals which artists have the most engaged, positive fanbase',
        'Use this to identify artists ready for major promotion pushes'
    ],
    caption='Divergent chart - positive sentiment builds up, negative goes down'
)
```

```python
# Create sentiment cluster analysis chart
cluster_chart = create_sentiment_cluster_chart(
    df=sentiment_df,
    sentiment_score_col='sentiment_score',
    category_col='sentiment_category',
    artist_col='artist_name',
    title='🎯 Sentiment Model in Action - Score Clustering'
)

story_block(
    fig=cluster_chart,
    title='🤖 Our Custom Sentiment Model at Work',
    bullets=[
        'Each dot represents a fan comment with its sentiment score',
        'Green dots = positive sentiment, Red dots = negative sentiment',
        'Clustering shows how well our model separates positive from negative',
        'Our music-trained model understands slang like "fire" and "mid"'
    ],
    caption='Vertical line at 0 = neutral sentiment boundary'
)
```

```python
# Extract top positive and negative comments with percentages
print('🌟 TOP 3 POSITIVE THINGS FANS SAY\n')

positive_comments = extract_top_positive_comments(
    df=sentiment_df,
    artist_col='artist_name',
    comment_col='comment',
    sentiment_col='sentiment_category',
    top_n=3
)

for artist, comments in positive_comments.items():
    print(f'**{artist}:**')
    for i, comment in enumerate(comments, 1):
        print(f'  {i}. "{comment}"')
    print()

print('\n⚠️ TOP 3 NEGATIVE FEEDBACK WITH PERCENTAGES\n')

negative_analysis = extract_top_negative_comments_with_percentages(
    df=sentiment_df,
    artist_col='artist_name',
    comment_col='comment',
    sentiment_col='sentiment_category',
    top_n=3
)

for artist, data in negative_analysis.items():
    pos_pct = data['positive_percentage']
    neg_pct = data['negative_percentage']
    
    print(f'**{artist}:** {pos_pct}% positive, {neg_pct}% negative')
    
    if data['negative_comments']:
        print('  Areas for improvement:')
        for i, comment in enumerate(data['negative_comments'], 1):
            print(f'    {i}. "{comment}"')
    else:
        print('  🎉 No significant negative feedback!')
    print()
```

```python
# Identify standout videos for experimentation
standout_videos = identify_standout_videos(
    df=video_df,
    views_col='views',
    positive_sentiment_col='positive_sentiment_pct',
    min_positive_threshold=80,  # High positive sentiment
    max_views_threshold=50000   # Normal view counts
)

print('🎯 STANDOUT VIDEOS FOR EXPERIMENTATION\n')
print('Videos with HIGH positive sentiment but NORMAL view counts:')
print('(Perfect for testing promotion strategies!)\n')

if len(standout_videos) > 0:
    for _, video in standout_videos.iterrows():
        print(f'🎬 {video["video_title"]}')
        print(f'   Views: {video["views"]:,} (normal range)')
        print(f'   Positive Sentiment: {video["positive_sentiment_pct"]:.1f}% (high!)')
        print(f'   💡 Experiment: Boost promotion and track view growth\n')
else:
    print('No videos found matching criteria. Try adjusting thresholds.')

print(f'Total standout videos identified: {len(standout_videos)}')
```

```python
# Roster-wide sentiment analysis and tour grouping
print('🎤 ROSTER-WIDE SENTIMENT ANALYSIS\n')

roster_analysis = analyze_roster_sentiment(
    df=sentiment_df,
    artist_col='artist_name',
    sentiment_col='sentiment_category',
    comment_col='comment'
)

for artist, analysis in roster_analysis.items():
    fan_type = analysis['fan_type']
    sentiment = analysis['sentiment_summary']
    compatibility = analysis['tour_compatibility']
    ratio = analysis['positive_ratio']
    
    print(f'**{artist}:**')
    print(f'  Fan Type: {fan_type.title()}')
    print(f'  Sentiment: {sentiment.title()} ({ratio:.1%} positive)')
    print(f'  Tour Style: {compatibility.replace("_", " ").title()}')
    print()

print('\n🎪 TOUR GROUPING RECOMMENDATIONS\n')

tour_groups = group_artists_for_tours(roster_analysis)

print('Artists grouped by fan compatibility:')
for compatibility, artists in tour_groups['tour_groups'].items():
    artist_names = [a['artist'] for a in artists]
    print(f'\n**{compatibility.replace("_", " ").title()} Tours:**')
    print(f'  Artists: {", ".join(artist_names)}')
    print(f'  Why: Similar fan energy levels and venue preferences')

print('\n**Recommendations:**')
for rec in tour_groups['recommendations']:
    print(f'  • {rec}')
```

```python
# Analyze music industry slang in comments
print('🔥 MUSIC SLANG ANALYSIS\n')

all_comments = sentiment_df['comment'].tolist()
slang_analysis = detect_music_slang(all_comments)

print(f'Total slang instances found: {slang_analysis["total_slang_instances"]}')
print(f'Unique slang terms: {", ".join(slang_analysis["slang_terms_found"])}')
print()
print('Slang sentiment impact:')
print(f'  Positive slang usage: {slang_analysis["slang_sentiment_impact"]["positive"]} times')
print(f'  Negative slang usage: {slang_analysis["slang_sentiment_impact"]["negative"]} times')
print()
print('Most frequent slang terms:')
for term, count in slang_analysis['term_frequency'].items():
    print(f'  "{term}": {count} times')
print()
print('💡 Why this matters:')
print('  • Our custom model understands music slang better than general models')
print('  • Slang detection helps identify authentic fan engagement')
print('  • Terms like "fire" and "slaps" are strong positive indicators')
print('  • "Mid" and "not it" are modern negative expressions')
```
