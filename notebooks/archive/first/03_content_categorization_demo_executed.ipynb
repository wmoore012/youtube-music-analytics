**Focus: Understanding what content strategies work best for each artist** 🎯

```python
# Import required modules
import pandas as pd
import numpy as np
from datetime import datetime, timedelta
import plotly.express as px
import plotly.graph_objects as go

# Import content analysis functions
from youtubeviz.content import (
    calculate_views_by_category,
    analyze_genre_diversity,
    categorize_content_types,
    classify_video_duration,
    analyze_isrc_distribution,
    calculate_content_performance_metrics,
    analyze_content_strategy_effectiveness,
    identify_content_opportunities
)

# Import chart functions
from youtubeviz.charts import (
    create_artist_strengths_venn_diagram,
    create_isrc_balance_chart,
    create_duration_breakdown_chart,
    create_content_type_breakdown_chart,
    create_artist_content_comparison_chart,
    create_roster_content_overview_chart
)

# Import storytelling functions
from youtubeviz.storytelling import story_block

print('✅ Content categorization toolkit loaded!')
```

```python
# ISRC vs Non-ISRC Balance Analysis
isrc_chart = create_isrc_balance_chart(
    df=df,
    artist_col='artist_name',
    isrc_col='has_isrc',
    views_col='views'
)

story_block(
    fig=isrc_chart,
    title='🎵 Music Content vs Content Videos Strategy',
    bullets=[
        'Green bars = Music content with ISRC (official releases)',
        'Red bars = Content videos without ISRC (lifestyle, behind-scenes)',
        'Flyana Boss dominates with lifestyle content (as noted in requirements)',
        'BiC Fizzle focuses heavily on official music videos',
        'Balance shows each artist\'s unique content strategy'
    ],
    caption='ISRC = International Standard Recording Code (indicates official music releases)'
)
```

```python
# Short-form vs Long-form Video Breakdown
duration_chart = create_duration_breakdown_chart(
    df=df,
    artist_col='artist_name',
    duration_col='duration_seconds',
    views_col='views',
    short_form_threshold=300  # 5 minutes
)

story_block(
    fig=duration_chart,
    title='⏱️ Short-form vs Long-form Content Performance',
    bullets=[
        'Blue bars = Short-form content (≤5 minutes) - optimized for social media',
        'Green bars = Long-form content (>5 minutes) - deeper engagement',
        'Most music videos are short-form for maximum shareability',
        'Lifestyle content tends to be longer for storytelling',
        'Strategy varies by artist personality and audience preferences'
    ],
    caption='Duration strategy impacts algorithm performance and audience retention'
)
```

```python
# Content Type Breakdown: Music Video vs Lyric Video vs Visualizer vs Other
content_breakdown_chart = create_content_type_breakdown_chart(
    df=df,
    artist_col='artist_name',
    content_type_col='content_type',
    views_col='views'
)

story_block(
    fig=content_breakdown_chart,
    title='📹 Content Type Analysis: Count vs Performance',
    bullets=[
        'Left chart shows video count by content type per artist',
        'Right chart shows total views by content type per artist',
        'Compare quantity vs quality - some artists get more views with fewer videos',
        'Music videos typically perform best but require most resources',
        'Lifestyle content can outperform music content (see Flyana Boss)'
    ],
    caption='Understanding which content types drive the most engagement per artist'
)
```

```python
# Side-by-Side Artist Comparison
comparison_chart = create_artist_content_comparison_chart(
    df=df,
    artist_col='artist_name',
    content_type_col='content_type',
    views_col='views',
    comparison_type='side_by_side'
)

story_block(
    fig=comparison_chart,
    title='🥊 Side-by-Side Artist Content Strategy Comparison',
    bullets=[
        'Each color represents a different artist in our roster',
        'Compare how each artist performs in different content categories',
        'Identify which artists excel in which content types',
        'Spot opportunities for artists to learn from each other',
        'Notice genre differences: electronic vs hip-hop vs R&B approaches'
    ],
    caption='Direct comparison reveals strengths and opportunities for each artist'
)
```

```python
# Combined Roster Analysis
roster_overview_chart = create_roster_content_overview_chart(
    df=df,
    artist_col='artist_name',
    content_type_col='content_type',
    views_col='views'
)

story_block(
    fig=roster_overview_chart,
    title='🎤 Combined Roster Content Strategy Overview',
    bullets=[
        'Shows overall content distribution across entire roster',
        'Reveals label-wide content strategy patterns',
        'Identifies which content types drive most total engagement',
        'Helps allocate resources across different content production areas',
        'Lifestyle content surprisingly dominates due to Flyana Boss success'
    ],
    caption='Pie chart shows percentage of total roster views by content type'
)
```

```python
# Genre Context Analysis (All New Signees, Different Genres)
print('🎵 GENRE DIVERSITY ANALYSIS\n')
print('**Context: All artists are new signees but represent different genres**\n')

genre_analysis = analyze_genre_diversity(
    df=df,
    artist_col='artist_name',
    genre_col='genre',
    content_type_col='content_type'
)

print(f'**Genre Diversity Score:** {genre_analysis["genre_diversity"]:.2f} (1.0 = maximum diversity)')
print(f'**Total Genres Represented:** {genre_analysis["unique_genres"]} across {genre_analysis["total_artists"]} artists\n')

print('**Artist Genre Breakdown:**')
for artist, genre in genre_analysis['artist_genres'].items():
    print(f'  • {artist}: {genre.title()}')

print(f'\n**Diversity Insight:** {genre_analysis["diversity_analysis"]}\n')

print('**Content Strategy by Genre:**')
content_by_genre = genre_analysis['content_strategy_by_genre']
for genre, content_types in content_by_genre.items():
    print(f'\n**{genre.title()} Artists:**')
    for content_type, count in content_types.items():
        if count > 0:
            print(f'  • {content_type.replace("_", " ").title()}: {count} videos')

print('\n💡 **Why Genre Diversity Matters:**')
print('  • Different genres attract different audience demographics')
print('  • Content strategies vary by genre (electronic vs hip-hop vs R&B)')
print('  • Cross-genre collaboration opportunities for unique content')
print('  • Diversified risk across multiple music market segments')
```

```python
# ISRC Distribution Analysis with Percentages
print('📊 ISRC vs NON-ISRC DETAILED ANALYSIS\n')

isrc_analysis = analyze_isrc_distribution(
    df=df,
    artist_col='artist_name',
    isrc_col='has_isrc',
    views_col='views'
)

print('**Balance Analysis: Videos with ISRC vs Without (Music vs Content Videos)**\n')

for _, row in isrc_analysis.iterrows():
    artist = row['artist_name']
    isrc_pct = row['isrc_percentage']
    non_isrc_pct = row['non_isrc_percentage']
    isrc_view_pct = row['isrc_view_percentage']
    
    print(f'**{artist}:**')
    print(f'  Video Count: {isrc_pct}% music content (ISRC), {non_isrc_pct}% content videos')
    print(f'  View Distribution: {isrc_view_pct}% from music, {100-isrc_view_pct:.1f}% from content')
    
    # Strategy assessment
    if isrc_view_pct > 70:
        strategy = 'Music-Focused Strategy'
    elif isrc_view_pct < 30:
        strategy = 'Content-Focused Strategy'
    else:
        strategy = 'Balanced Strategy'
    
    print(f'  Strategy Type: {strategy}')
    print()

print('💡 **Strategic Insights:**')
print('  • ISRC videos = Official music releases (albums, singles, EPs)')
print('  • Non-ISRC videos = Content marketing (lifestyle, behind-scenes, vlogs)')
print('  • Flyana Boss proves content videos can outperform music videos')
print('  • BiC Fizzle shows traditional music-focused approach still works')
print('  • Strategy should match artist personality and audience expectations')
```

```python
# Content Performance Metrics Summary
print('📈 CONTENT PERFORMANCE METRICS SUMMARY\n')

performance_metrics = calculate_content_performance_metrics(
    df=df,
    content_type_col='content_type',
    views_col='views'
)

print('**Performance by Content Type (Ranked by Total Views):**\n')
display(performance_metrics)

print('\n**Key Insights:**')
top_performer = performance_metrics.iloc[0]
print(f'  • **Top Performer**: {top_performer["content_type"]} ({top_performer["view_share_percentage"]}% of total views)')
print(f'  • **Most Efficient**: {performance_metrics.loc[performance_metrics["avg_views"].idxmax(), "content_type"]} (highest avg views per video)')
print(f'  • **Most Prolific**: {performance_metrics.loc[performance_metrics["video_count"].idxmax(), "content_type"]} (most videos produced)')

print('\n💡 **Content Strategy Recommendations:**')
for _, row in performance_metrics.head(3).iterrows():
    content_type = row['content_type']
    avg_views = row['avg_views']
    print(f'  • **{content_type.replace("_", " ").title()}**: Avg {avg_views:,.0f} views - {"High priority" if avg_views > 400000 else "Moderate priority" if avg_views > 200000 else "Experimental"} content type')
```
