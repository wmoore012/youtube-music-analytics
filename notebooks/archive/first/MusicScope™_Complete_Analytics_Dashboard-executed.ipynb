# 🔧 Setup & Data Generation {#setup}

Loading all analytics modules and generating sample data for demonstration.

```python
# 🎵 MusicScope™ - Complete Analytics Toolkit
import pandas as pd
import numpy as np
from datetime import datetime, timedelta
import plotly.express as px
import plotly.graph_objects as go

# Import all our custom modules
from youtubeviz.storytelling import story_block, quick_takeaways, narrative_intro
from youtubeviz.charts import (
    views_over_time_plotly, enhance_chart_beauty,
    create_divergent_sentiment_chart, create_sentiment_cluster_chart
)
from youtubeviz.sentiment import (
    extract_top_positive_comments, extract_top_negative_comments_with_percentages,
    identify_standout_videos, analyze_roster_sentiment, group_artists_for_tours
)

print('🎯 MusicScope™ Complete Analytics Toolkit Loaded!')
print('✅ ChartFlow™: Performance charts ready')
print('✅ SentimentScope™: Fan insights ready')
print('✅ ContentFlow™: Video analysis ready')
```

```python
# Generate comprehensive sample data for all 4 artists
artists = ['Flyana Boss', 'BiC Fizzle', 'COBRAH', 're6ce']
dates = pd.date_range(end=datetime.now(), periods=30, freq='D')

# Performance data
performance_data = []
for artist in artists:
    for date in dates:
        base_views = np.random.randint(10000, 100000)
        views = base_views + np.random.randint(-5000, 15000)
        likes = int(views * np.random.uniform(0.02, 0.08))
        comments = int(views * np.random.uniform(0.005, 0.02))
        
        performance_data.append({
            'artist_name': artist,
            'date': date,
            'views': max(views, 1000),
            'likes': max(likes, 10),
            'comments': max(comments, 5),
            'engagement_rate': (likes + comments) / max(views, 1) * 100
        })

df = pd.DataFrame(performance_data)
print(f'📊 Generated {len(df)} performance records for {len(artists)} artists')
print(f'📅 Date range: {df.date.min().date()} to {df.date.max().date()}')
print(f'🎤 Artists: {", ".join(artists)}')
print(f'🔍 Unique artists in data: {df.artist_name.nunique()}')
df.head()
```

---
# 📊 ChartFlow™ - Performance Analytics {#chartflow}

Interactive charts showing artist performance, engagement metrics, and growth trends.

```python
# ChartFlow™ - Views Over Time Analysis
fig = views_over_time_plotly(
    df=df,
    date_col='date',
    value_col='views',
    group_col='artist_name'
)
fig.update_layout(title='📈 ChartFlow™ - YouTube Views Over Time')

story_block(
    fig=fig,
    title='🎯 Artist Performance Trends',
    bullets=[
        f'Tracking {df.artist_name.nunique()} artists over {len(df.date.unique())} days',
        'Interactive chart - hover for details, click legend to filter',
        'Identify momentum shifts and viral moments',
        'Compare growth trajectories between artists'
    ],
    caption='ChartFlow™ - Performance visualization engine'
)
```

```python
# ChartFlow™ - Engagement Rate Analysis
avg_engagement = df.groupby('artist_name')['engagement_rate'].mean().reset_index()
avg_engagement = avg_engagement.sort_values('engagement_rate', ascending=False)

fig2 = px.bar(
    avg_engagement,
    x='artist_name',
    y='engagement_rate',
    title='🔥 ChartFlow™ - Average Engagement Rate by Artist',
    color='engagement_rate',
    color_continuous_scale='viridis'
)
fig2.update_layout(showlegend=False)
fig2_enhanced = enhance_chart_beauty(fig2, title='🔥 ChartFlow™ - Fan Engagement Champions')

story_block(
    fig=fig2_enhanced,
    title='💬 Who Drives the Most Fan Interaction?',
    bullets=[
        f'{avg_engagement.iloc[0]["artist_name"]} leads with {avg_engagement.iloc[0]["engagement_rate"]:.1f}% engagement',
        'Higher engagement = stronger fan connection',
        'Focus marketing spend on high-engagement artists',
        'Engagement rate matters more than raw view counts'
    ],
    caption='ChartFlow™ - Engagement = (Likes + Comments) / Views × 100'
)
```

# 🎭 SentimentScope™ - Fan Insights {#sentimentscope}

Sentiment analysis, fan comment insights, and tour planning recommendations.

```python
# SentimentScope™ - Generate sample sentiment data
positive_comments = [
    'This track is absolute fire! 🔥', 'No cap, this is a banger',
    'Love the energy, this slaps hard', 'Amazing vocals, gives me chills'
]
negative_comments = [
    'Not feeling this one, mid tbh', 'Could be better, not my style',
    'The beat is off, doesn\'t work', 'Skip for me, not it'
]

sentiment_data = []
for artist in artists:
    # More positive than negative (realistic)
    for i in range(np.random.randint(15, 25)):
        sentiment_data.append({
            'artist_name': artist,
            'comment': np.random.choice(positive_comments),
            'sentiment_score': np.random.uniform(0.6, 0.95),
            'sentiment_category': 'positive'
        })
    for i in range(np.random.randint(3, 8)):
        sentiment_data.append({
            'artist_name': artist,
            'comment': np.random.choice(negative_comments),
            'sentiment_score': np.random.uniform(-0.8, -0.2),
            'sentiment_category': 'negative'
        })

sentiment_df = pd.DataFrame(sentiment_data)
print(f'🎭 SentimentScope™ - Generated {len(sentiment_df)} comments for {sentiment_df.artist_name.nunique()} artists')
sentiment_df.head()
```

```python
# SentimentScope™ - Divergent Sentiment Chart
divergent_chart = create_divergent_sentiment_chart(
    df=sentiment_df,
    artist_col='artist_name',
    sentiment_col='sentiment_category'
)
divergent_chart.update_layout(title='🎭 SentimentScope™ - Fan Sentiment Breakdown')

story_block(
    fig=divergent_chart,
    title='💬 What Fans Are Really Saying',
    bullets=[
        'Green bars show positive fan sentiment (above baseline)',
        'Red bars show areas for improvement (below baseline)',
        'Reveals which artists have the most engaged, positive fanbase',
        'Use this to identify artists ready for major promotion pushes'
    ],
    caption='SentimentScope™ - Divergent chart shows sentiment balance'
)
```

# 🎬 ContentFlow™ - Video Analysis {#contentflow}

Video categorization, content strategy, and performance optimization.

```python
# ContentFlow™ - Video categorization analysis
video_types = ['Music Video', 'Lyric Video', 'Behind Scenes', 'Live Performance', 'Visualizer']
content_data = []

for artist in artists:
    for video_type in video_types:
        # Generate realistic video performance data
        video_count = np.random.randint(1, 4)
        total_views = np.random.randint(50000, 500000)
        has_isrc = video_type in ['Music Video', 'Lyric Video']  # Music content has ISRC
        
        content_data.append({
            'artist_name': artist,
            'video_type': video_type,
            'video_count': video_count,
            'total_views': total_views,
            'has_isrc': has_isrc,
            'avg_views_per_video': total_views / video_count
        })

content_df = pd.DataFrame(content_data)
print(f'🎬 ContentFlow™ - Generated content analysis for {content_df.artist_name.nunique()} artists')
print(f'📊 Video types analyzed: {", ".join(video_types)}')
content_df.head()
```

```python
# ContentFlow™ - ISRC vs Non-ISRC Analysis
isrc_analysis = content_df.groupby(['artist_name', 'has_isrc']).agg({
    'video_count': 'sum',
    'total_views': 'sum'
}).reset_index()
isrc_analysis['content_type'] = isrc_analysis['has_isrc'].map({True: 'Music Content (ISRC)', False: 'Other Content'})

fig3 = px.bar(
    isrc_analysis,
    x='artist_name',
    y='total_views',
    color='content_type',
    title='🎬 ContentFlow™ - Music vs Content Video Performance',
    barmode='group'
)

story_block(
    fig=fig3,
    title='🎵 Music Content vs Other Content Strategy',
    bullets=[
        'Music content (ISRC) includes official music videos and lyric videos',
        'Other content includes behind-scenes, live performances, visualizers',
        'Balance shows artist content strategy and fan preferences',
        'Helps optimize content mix for maximum engagement'
    ],
    caption='ContentFlow™ - Content categorization and strategy analysis'
)
```
