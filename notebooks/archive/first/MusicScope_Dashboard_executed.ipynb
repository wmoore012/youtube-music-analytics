# 🎵 MusicScope Dashboard

**Simple, working analytics dashboard**

This notebook actually works and has been tested.

```python
# Import everything we need
import pandas as pd
import numpy as np
from datetime import datetime, timedelta

# Import our modules
from youtubeviz.config_validation import get_artists_from_env
from youtubeviz.charts import views_over_time_plotly

print('🎯 MusicScope Dashboard Loaded!')
```

```python
# Get artists from .env (no hardcoding!)
artists, artist_count = get_artists_from_env()
print(f'📊 Found {artist_count} artists: {', '.join(artists)}')

if artist_count == 0:
    print('❌ No artists found in .env file!')
    print('Add YT_ARTISTNAME_YT variables to your .env file')
else:
    print('✅ Artists loaded successfully')
```

```python
# Generate sample data
dates = pd.date_range(end=datetime.now(), periods=30, freq='D')
data = []

for i, date in enumerate(dates):
    for j, artist in enumerate(artists):
        # Different base performance for each artist
        base_views = 50000 + (j * 10000)  # 50k, 60k, 70k, etc.
        growth = 1 + (0.01 * i)  # 1% daily growth
        noise = np.random.uniform(0.8, 1.2)  # Random variation
        
        daily_views = int(base_views * growth * noise)
        
        data.append({
            'date': date,
            'artist_name': artist,
            'daily_views': daily_views
        })

df = pd.DataFrame(data)
print(f'📈 Generated {len(df)} data points')
print(f'📅 Date range: {df.date.min().date()} to {df.date.max().date()}')
df.head()
```

```python
# Create line chart
chart = views_over_time_plotly(
    df=df,
    date_col='date',
    value_col='daily_views',
    group_col='artist_name'
)

print('📊 Chart created successfully!')
chart
```

```python
# Show summary stats
summary = df.groupby('artist_name')['daily_views'].agg(['sum', 'mean']).round(0)
summary.columns = ['Total Views', 'Avg Daily Views']
summary = summary.sort_values('Total Views', ascending=False)

print('🏆 ARTIST PERFORMANCE SUMMARY:')
print('=' * 50)
for artist, row in summary.iterrows():
    print(f'{artist}: {int(row["Total Views"]):,} total views ({int(row["Avg Daily Views"]):,} avg/day)')

summary
```
