# 🔧 Setup & Data Generation

Loading all analytics modules and generating sample data for demonstration.

```python
# 🎵 MusicScope™ - Complete Analytics Toolkit
import pandas as pd
import numpy as np
from datetime import datetime, timedelta
import plotly.express as px
import plotly.graph_objects as go

# Import all our custom modules - TESTED AND WORKING
from youtubeviz.storytelling import story_block, quick_takeaways, narrative_intro
from youtubeviz.charts import (
    views_over_time_plotly, enhance_chart_beauty,
    create_divergent_sentiment_chart, create_sentiment_cluster_chart,
    create_isrc_balance_chart, create_content_type_breakdown_chart
)
from youtubeviz.content import create_artist_comparison_chart
from youtubeviz.sentiment import (
    extract_top_positive_comments, extract_top_negative_comments_with_percentages,
    identify_standout_videos, analyze_roster_sentiment
)
from youtubeviz.config_validation import (
    get_artists_from_env, validate_artist_count_in_data, 
    print_validation_results, EXPECTED_ARTIST_COUNT, EXPECTED_ARTISTS
)
from youtubeviz.summary_generator import (
    generate_executive_summary, create_actionable_recommendations
)

print('🎯 MusicScope™ Complete Analytics Toolkit Loaded!')
print('✅ ChartFlow™: Performance charts ready')
print('✅ SentimentScope™: Fan insights ready')
print('✅ ContentFlow™: Video analysis ready')
print('✅ Auto-Summary: Intelligent insights ready')
print('🎵 Dynamic Artist Support: Works with any number of artists from .env')
```

```python
# 🎯 DYNAMIC ARTIST CONFIGURATION: Get artists from .env
artists, artist_count = get_artists_from_env()
print(f'📊 Generating sample data for {artist_count} artists: {', '.join(artists)}')

# Validate we have artists configured
if artist_count == 0:
    print('🚨 ERROR: No artists found in .env configuration!')
    print('Please add YT_ARTISTNAME_YT variables to your .env file')
    raise ValueError('No artists configured')

dates = pd.date_range(end=datetime.now(), periods=30, freq='D')

# Dynamically generate artist profiles based on .env configuration
def generate_artist_profiles(artists):
    """Generate realistic performance profiles for any number of artists."""
    profiles = {}
    
    # Base performance tiers to distribute artists across
    performance_tiers = [
        {'base_views': 85000, 'growth_rate': 0.03, 'volatility': 0.2, 'tier': 'Strong Performer'},
        {'base_views': 75000, 'growth_rate': 0.025, 'volatility': 0.12, 'tier': 'High Performer'},
        {'base_views': 65000, 'growth_rate': 0.02, 'volatility': 0.15, 'tier': 'Solid Growth'},
        {'base_views': 55000, 'growth_rate': 0.01, 'volatility': 0.18, 'tier': 'Stable'},
        {'base_views': 45000, 'growth_rate': 0.05, 'volatility': 0.25, 'tier': 'Fast Growth'},
        {'base_views': 35000, 'growth_rate': -0.01, 'volatility': 0.3, 'tier': 'Needs Attention'}
    ]
    
    # Assign profiles to artists, cycling through tiers if more artists than tiers
    for i, artist in enumerate(artists):
        tier_index = i % len(performance_tiers)
        profiles[artist] = performance_tiers[tier_index].copy()
    
    return profiles

artist_profiles = generate_artist_profiles(artists)
print(f'📊 Generated performance profiles for {len(artist_profiles)} artists')
for artist, profile in artist_profiles.items():
    print(f'   {artist}: {profile["tier"]} (Base: {profile["base_views"]:,} views, Growth: {profile["growth_rate"]*100:+.1f}%)')
```

```python
# 📊 PERFORMANCE DATA GENERATION
performance_data = []
for i, date in enumerate(dates):
    for artist in artists:
        profile = artist_profiles[artist]
        
        # Calculate trending views based on growth rate and time
        base_trend = profile['base_views'] * (1 + profile['growth_rate']) ** i
        
        # Add realistic daily volatility
        daily_noise = np.random.normal(0, profile['volatility'])
        daily_views = int(base_trend * (1 + daily_noise))
        
        # Ensure views don't go negative
        daily_views = max(daily_views, 1000)
        
        performance_data.append({
            'date': date,
            'artist_name': artist,
            'daily_views': daily_views,
            'engagement_rate': np.random.uniform(0.02, 0.08),
            'tier': profile['tier']
        })

performance_df = pd.DataFrame(performance_data)
print(f'📈 Generated {len(performance_df)} performance data points')
print(f'📅 Date range: {performance_df.date.min().strftime("%Y-%m-%d")} to {performance_df.date.max().strftime("%Y-%m-%d")}')
performance_df.head()
```

# 📊 ChartFlow™ - Performance Analytics

Interactive performance charts showing artist trends, comparisons, and growth patterns.

```python
# 📈 LINE CHART: Views Over Time (Trending Performance) - FIXED PARAMETERS
line_chart = views_over_time_plotly(
    df=performance_df,
    date_col='date',
    value_col='daily_views',
    group_col='artist_name'
)

story_block(
    fig=line_chart,
    title='📈 Performance Trends: Who\'s Rising?',
    bullets=[
        'Line chart shows daily view trends over the past 30 days',
        'Upward trending lines indicate growing momentum',
        'Steep slopes reveal which artists are gaining traction fastest',
        'Use this to identify artists ready for increased marketing investment'
    ],
    caption='Interactive chart - hover for exact values, click legend to filter artists'
)
```

```python
# 📊 BAR CHART: Artist Performance Comparison - FIXED PARAMETERS
# Calculate total views and average engagement for comparison
artist_summary = performance_df.groupby('artist_name').agg({
    'daily_views': ['sum', 'mean'],
    'engagement_rate': 'mean',
    'tier': 'first'
}).round(2)

artist_summary.columns = ['total_views', 'avg_daily_views', 'avg_engagement', 'tier']
artist_summary = artist_summary.reset_index()

comparison_chart = create_artist_comparison_chart(
    df=artist_summary,
    artist_col='artist_name',
    metrics_cols=['total_views']
)

story_block(
    fig=comparison_chart,
    title='🏆 Who\'s Dominating the Numbers?',
    bullets=[
        'Bar chart compares total views across all artists',
        'Taller bars indicate higher overall performance',
        'Use this to identify your top performers and rising stars',
        'Consider cross-promotion opportunities between high and emerging artists'
    ],
    caption='Total views accumulated over the past 30 days'
)

print('📊 Artist Performance Summary:')
for _, row in artist_summary.iterrows():
    print(f'   {row["artist_name"]}: {row["total_views"]:,} total views ({row["tier"]})')
```
