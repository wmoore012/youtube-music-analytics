## 📊 Data Loading & Preparation

Let's create some realistic sample data to demonstrate our storytelling capabilities. In a real scenario, this would connect to your YouTube analytics database.

```python
# Create sample data that tells an interesting story
np.random.seed(42)  # For reproducible results

# Our featured artists
artists = ["Flyana Boss", "COBRAH", "Raiche"]
colors = {"Flyana Boss": "#FF6B6B", "COBRAH": "#4ECDC4", "Raiche": "#45B7D1"}

# Generate 30 days of data
dates = pd.date_range(start=datetime.now() - timedelta(days=30), end=datetime.now(), freq='D')

# Create realistic performance data with different growth patterns
data = []
for artist in artists:
    base_views = {"Flyana Boss": 50000, "COBRAH": 35000, "Raiche": 42000}[artist]
    growth_rate = {"Flyana Boss": 0.03, "COBRAH": 0.01, "Raiche": 0.02}[artist]
    
    for i, date in enumerate(dates):
        # Add growth trend + some randomness + weekend boost
        weekend_boost = 1.3 if date.weekday() >= 5 else 1.0
        trend_multiplier = (1 + growth_rate) ** i
        noise = np.random.normal(1, 0.15)
        
        views = int(base_views * trend_multiplier * weekend_boost * noise)
        likes = int(views * np.random.uniform(0.02, 0.05))  # 2-5% like rate
        comments = int(views * np.random.uniform(0.001, 0.003))  # 0.1-0.3% comment rate
        
        data.append({
            'date': date,
            'artist_name': artist,
            'views': views,
            'likes': likes,
            'comments': comments,
            'engagement_rate': ((likes + comments) / views) * 100
        })

df = pd.DataFrame(data)
print(f"📈 Generated {len(df)} data points across {len(artists)} artists")
print(f"📅 Date range: {df['date'].min().strftime('%Y-%m-%d')} to {df['date'].max().strftime('%Y-%m-%d')}")
df.head()
```

## 🏆 Performance Overview

Let's start with the big picture - how are our artists performing overall?

```python
# Calculate summary metrics for each artist
summary_stats = df.groupby('artist_name').agg({
    'views': ['sum', 'mean'],
    'likes': 'sum',
    'comments': 'sum',
    'engagement_rate': 'mean'
}).round(2)

# Flatten column names
summary_stats.columns = ['total_views', 'avg_daily_views', 'total_likes', 'total_comments', 'avg_engagement_rate']
summary_stats = summary_stats.reset_index()

# Create a beautiful bar chart
fig = px.bar(
    summary_stats, 
    x='artist_name', 
    y='total_views',
    color='artist_name',
    color_discrete_map=colors,
    title="Total Views - Last 30 Days"
)

# Enhance the chart beauty
fig = enhance_chart_beauty(
    fig,
    title="🎯 Total Views Showdown",
    emotional_theme="energetic"
)

# Create insights for our story block
top_performer = summary_stats.loc[summary_stats['total_views'].idxmax(), 'artist_name']
top_views = summary_stats.loc[summary_stats['total_views'].idxmax(), 'total_views']

bullets = [
    f"{top_performer} leads with {top_views:,.0f} total views",
    "All artists show strong weekend performance patterns",
    "Consistent daily posting drives cumulative growth",
    "View counts correlate with engagement quality"
]

story_block(
    fig=fig,
    title="🔥 Who's Dominating the View Count?",
    bullets=bullets,
    caption="Action: Focus marketing budget on momentum leaders"
)

print("\n📊 Summary Statistics:")
display(summary_stats)
```

## 📈 Engagement Deep Dive

Views are great, but engagement tells us about fan loyalty and content quality.

```python
# Create an engagement rate comparison
fig = px.scatter(
    df,
    x='views',
    y='engagement_rate',
    color='artist_name',
    color_discrete_map=colors,
    size='comments',
    hover_data=['date', 'likes'],
    title="Engagement Rate vs Views"
)

# Enhance with professional theme
fig = enhance_chart_beauty(
    fig,
    title="💎 Quality vs Quantity: The Engagement Story",
    emotional_theme="professional"
)

# Calculate engagement insights
avg_engagement = df.groupby('artist_name')['engagement_rate'].mean().sort_values(ascending=False)
engagement_leader = avg_engagement.index[0]
engagement_rate = avg_engagement.iloc[0]

bullets = [
    f"{engagement_leader} has the highest engagement at {engagement_rate:.1f}%",
    "Higher engagement often predicts long-term success",
    "Bubble size shows comment volume - fan conversation depth",
    "Sweet spot: High views + High engagement = Investment priority"
]

story_block(
    fig=fig,
    title="🎯 Finding the Engagement Champions",
    bullets=bullets,
    caption="Insight: Engagement quality matters more than raw view counts"
)
```

### 📚 What's Engagement Rate?

Engagement rate measures how actively fans interact with content beyond just watching. It includes likes, comments, shares, and saves divided by total views.

*Think of it like applause at a concert - views are attendance, but engagement shows how much the audience loved the show!*

**Industry Benchmark:** 2-4% is solid, 5%+ is exceptional for established artists.

## 📈 Momentum Analysis

Let's see who's trending upward and who might need some strategic support.

```python
# Create a time series showing momentum
fig = px.line(
    df,
    x='date',
    y='views',
    color='artist_name',
    color_discrete_map=colors,
    title="Views Over Time - Momentum Tracking"
)

# Add trend lines
for artist in artists:
    artist_data = df[df['artist_name'] == artist].copy()
    artist_data['day_number'] = range(len(artist_data))
    
    # Calculate trend line
    z = np.polyfit(artist_data['day_number'], artist_data['views'], 1)
    p = np.poly1d(z)
    
    fig.add_trace(go.Scatter(
        x=artist_data['date'],
        y=p(artist_data['day_number']),
        mode='lines',
        name=f'{artist} Trend',
        line=dict(dash='dash', color=colors[artist]),
        showlegend=False
    ))

# Enhance with warm theme for momentum story
fig = enhance_chart_beauty(
    fig,
    title="🚀 Momentum Tracker: Who's Rising?",
    emotional_theme="warm"
)

# Calculate momentum metrics
momentum_data = []
for artist in artists:
    artist_df = df[df['artist_name'] == artist].sort_values('date')
    recent_avg = artist_df.tail(7)['views'].mean()
    earlier_avg = artist_df.head(7)['views'].mean()
    momentum_pct = ((recent_avg - earlier_avg) / earlier_avg) * 100
    momentum_data.append({'artist': artist, 'momentum': momentum_pct})

momentum_df = pd.DataFrame(momentum_data).sort_values('momentum', ascending=False)
top_momentum = momentum_df.iloc[0]

bullets = [
    f"{top_momentum['artist']} shows strongest momentum: +{top_momentum['momentum']:.1f}%",
    "Dashed lines show overall trend direction",
    "Weekend spikes visible across all artists",
    "Consistent growth beats viral spikes for long-term success"
]

story_block(
    fig=fig,
    title="📊 Reading the Momentum Signals",
    bullets=bullets,
    caption="Strategy: Double down on artists with positive momentum trends"
)

print("\n🚀 Momentum Rankings:")
for _, row in momentum_df.iterrows():
    direction = "📈" if row['momentum'] > 0 else "📉"
    print(f"{direction} {row['artist']}: {row['momentum']:+.1f}%")
```

### 🚀 Understanding Momentum

Momentum tracks how fast an artist's metrics are changing. Positive momentum means growing views, subscribers, or engagement. It's like measuring if a song is climbing or falling on the charts.

*Momentum matters more than absolute numbers for investment decisions!*

**Key Insight:** Consistent 10% monthly growth often outperforms sporadic viral spikes.

## 🎯 Strategic Recommendations

Based on our data detective work, here are the actionable insights for your music industry decisions:

```python
# Generate personalized recommendations for each artist
recommendations = {}

for artist in artists:
    artist_data = summary_stats[summary_stats['artist_name'] == artist].iloc[0]
    momentum = momentum_df[momentum_df['artist'] == artist]['momentum'].iloc[0]
    
    # Determine recommendation based on performance
    if momentum > 15 and artist_data['avg_engagement_rate'] > 3:
        rec_type = "🚀 PRIORITY INVESTMENT"
        actions = [
            "Increase marketing budget by 40%",
            "Fast-track next single release",
            "Book premium playlist placements",
            "Consider collaboration opportunities"
        ]
    elif momentum > 5:
        rec_type = "📈 GROWTH OPPORTUNITY"
        actions = [
            "Maintain current promotion level",
            "Test new content formats",
            "Analyze successful content patterns",
            "Gradual budget increase"
        ]
    else:
        rec_type = "🔄 STRATEGIC PIVOT"
        actions = [
            "Analyze content strategy gaps",
            "A/B test different approaches",
            "Focus on engagement over reach",
            "Consider rebranding elements"
        ]
    
    recommendations[artist] = {
        'type': rec_type,
        'actions': actions,
        'momentum': momentum,
        'engagement': artist_data['avg_engagement_rate']
    }

# Display recommendations in a beautiful format
print("🎯 STRATEGIC RECOMMENDATIONS\n" + "="*50)

for artist, rec in recommendations.items():
    print(f"\n🎤 {artist.upper()}")
    print(f"Status: {rec['type']}")
    print(f"Momentum: {rec['momentum']:+.1f}% | Engagement: {rec['engagement']:.1f}%")
    print("\nRecommended Actions:")
    for i, action in enumerate(rec['actions'], 1):
        print(f"  {i}. {action}")
    print("-" * 40)
```
