## 📊 Quick Analytics


```python
# 📦 Import essential packages
import pandas as pd
import numpy as np
import plotly.express as px
import plotly.graph_objects as go
from datetime import datetime, timedelta

print('📊 Analytics packages loaded successfully!')
print(f'🕐 Generated at: {datetime.now()}')
```

```python
# 🎯 Sample analytics data
sample_data = pd.DataFrame({
    'date': pd.date_range('2024-01-01', periods=30, freq='D'),
    'views': np.random.randint(1000, 10000, 30),
    'engagement': np.random.uniform(0.02, 0.08, 30)
})

print(f'📈 Generated {len(sample_data)} days of sample data')
sample_data.head()
```

```python
# 📊 Create visualization
fig = px.line(sample_data, x='date', y='views', 
              title='📈 Views Over Time',
              labels={'views': 'Daily Views', 'date': 'Date'})

fig.update_layout(
    template='plotly_white',
    height=400
)

fig.show()
print('✅ Chart created successfully!')
```
