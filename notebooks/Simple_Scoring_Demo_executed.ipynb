```python
# 🚀 Setup and Imports
import sys
import pandas as pd
import numpy as np
from datetime import datetime, timedelta
import warnings
warnings.filterwarnings('ignore')

# Add project root to path
sys.path.insert(0, '..')

print('🚀 Basic imports completed')
print('📊 Ready for real data analysis')
```

```python
# 📊 Load Real Data
from youtubeviz.data import load_artist_daily_metrics
from web.etl_helpers import get_engine

# Load real data from database
engine = get_engine()
end_date = datetime.now().date()
start_date = end_date - timedelta(days=30)

print(f'📅 Loading data from {start_date} to {end_date}')

# Load real artist data
real_data = load_artist_daily_metrics(
    start=start_date,
    end=end_date,
    engine=engine
)

print(f'✅ REAL DATA LOADED: {len(real_data)} records')
print(f'📊 Artists: {real_data["artist_name"].nunique()}')
print(f'🎵 Videos: {real_data["video_id"].nunique()}')

# Show sample of real data
print('\n📋 Sample Real Data:')
display(real_data.head())
```

```python
# 🎯 Initialize Scoring System
from src.data_organization.scoring_engine import ScoringEngine
from src.data_organization.youtube_scoring_plugins import (
    ArtistMomentumScoringPlugin,
    EngagementScoringPlugin,
    GrowthPotentialScoringPlugin
)

# Create scoring engine with storage
scoring_engine = ScoringEngine(enable_storage=True)

# Register scoring plugins
momentum_plugin = ArtistMomentumScoringPlugin()
engagement_plugin = EngagementScoringPlugin()
growth_plugin = GrowthPotentialScoringPlugin()

scoring_engine.register_plugin(momentum_plugin)
scoring_engine.register_plugin(engagement_plugin)
scoring_engine.register_plugin(growth_plugin)

print('✅ Scoring engine initialized')
print(f'🔧 Available algorithms: {scoring_engine.get_available_algorithms()}')

# Show system status
status = scoring_engine.get_system_status()
print(f'📊 System Status:')
print(f'   - Loaded plugins: {status["loaded_plugins"]}')
print(f'   - Storage enabled: {status["storage_enabled"]}')
print(f'   - Schema valid: {status.get("storage_schema_valid", "Unknown")}')
```

```python
# 🚀 Artist Momentum Scoring with Real Data
if not real_data.empty:
    # Prepare data for momentum scoring
    momentum_data = real_data.copy()
    momentum_data = momentum_data.rename(columns={
        'date': 'metrics_date',
        'views': 'view_count',
        'likes': 'like_count',
        'comments': 'comment_count'
    })
    
    # Add required columns
    momentum_data['published_at'] = momentum_data['metrics_date']
    momentum_data['channel_title'] = momentum_data['artist_name']
    
    print('🎯 Running Artist Momentum Scoring...')
    
    # Execute momentum scoring
    momentum_result = scoring_engine.execute_scoring(
        'artist_momentum_scorer',
        momentum_data,
        store_results=True,
        entity_type='artist'
    )
    
    print(f'✅ Momentum scoring completed!')
    print(f'📊 Results: {len(momentum_result.entity_scores)} artists scored')
    print(f'💾 Run ID: {momentum_result.metadata.get("run_id", "N/A")}')
    
    # Display results
    momentum_scores = momentum_result.entity_scores.sort_values('score_value', ascending=False)
    print('\n🏆 Artist Momentum Rankings:')
    display(momentum_scores[['entity_id', 'score_value', 'confidence', 'momentum_category']])
    
else:
    print('❌ No real data available for momentum scoring')
```

```python
# 📈 Growth Potential Scoring with Real Data
if not real_data.empty:
    # Prepare data for growth scoring
    growth_data = real_data.copy()
    growth_data = growth_data.rename(columns={
        'date': 'metrics_date',
        'views': 'view_count',
        'likes': 'like_count',
        'comments': 'comment_count'
    })
    
    print('📈 Running Growth Potential Scoring...')
    
    # Execute growth scoring
    growth_result = scoring_engine.execute_scoring(
        'growth_potential_scorer',
        growth_data,
        store_results=True,
        entity_type='artist'
    )
    
    print(f'✅ Growth potential scoring completed!')
    print(f'📊 Results: {len(growth_result.entity_scores)} artists analyzed')
    print(f'💾 Run ID: {growth_result.metadata.get("run_id", "N/A")}')
    
    # Display results
    growth_scores = growth_result.entity_scores.sort_values('score_value', ascending=False)
    print('\n🚀 Growth Potential Rankings:')
    display(growth_scores[['entity_id', 'score_value', 'confidence', 'trend_direction']])
    
else:
    print('❌ No real data available for growth scoring')
```

```python
# 📊 Query Stored Scoring Results
from src.data_organization.scoring_storage import ScoringStorage

storage = ScoringStorage()

print('📊 Querying Stored Scoring Results...')

# Get latest scores
latest_scores = storage.get_latest_scores(
    algorithm_name='artist_momentum_scorer',
    entity_type='artist',
    limit=10
)

if not latest_scores.empty:
    print(f'✅ Found {len(latest_scores)} stored results')
    print('\n📋 Latest Momentum Scores from Database:')
    display(latest_scores[['entity_id', 'score_value', 'confidence_level', 'calculation_timestamp']])
else:
    print('❌ No stored results found')

# Get algorithm performance
performance = storage.get_algorithm_performance()
if not performance.empty:
    print('\n🔧 Algorithm Performance:')
    display(performance[['algorithm_name', 'total_runs', 'total_results', 'overall_avg_score']])
else:
    print('❌ No performance data available')
```
