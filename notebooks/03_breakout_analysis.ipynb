```python
# Notebook bootstrap: ensure project root is on sys.path for `import src.*`
import os, sys
from pathlib import Path
nb_cwd = Path.cwd()
root = None
for p in [nb_cwd, *nb_cwd.parents]:
    if (p / 'pyproject.toml').exists() or (p / 'src').exists():
        root = p; break
if root and str(root) not in sys.path:
    sys.path.insert(0, str(root))
os.environ['PYTHONPATH'] = str(root) + (':' + os.environ.get('PYTHONPATH','') if os.environ.get('PYTHONPATH') else '')
print(f'Project root: {root}')
```

```python
import pandas as pd
import plotly.express as px
from src.config import load_settings
from src import qa as _qa
from src.storytelling import so_what

settings = load_settings()
with _qa.time_limit(settings.execution_timeout_s):
    scores_path = settings.processed_dir / 'artist_breakout_scores.csv'
    df = pd.read_csv(scores_path, parse_dates=['last_metrics_date'])
    top = df.sort_values('composite_score', ascending=False).head(6).copy()
    top['label'] = top['artist_name'] + ' — ' + (top['composite_score'].round(1)).astype(str)

with _qa.time_limit(settings.execution_timeout_s):
    fig = px.bar(top, x='artist_name', y='composite_score', color='decision',
                 title='Breakout candidates — prioritize outreach (scores 0–100)')
    fig.update_layout(yaxis_title='Composite Score', xaxis_title='Artist', template='plotly_white')
fig.show()

# So-What: one-line implication
so_what('Top 2–3 artists exceed thresholds; prioritize A&R touchpoints this week.')
```
